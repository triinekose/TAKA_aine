Järgmine lahter otsib faili repo `data` kaustast. Kui kohalikku faili pole, loetakse sama andmestik kursuse repo veebiaadressilt. Lahter kuvab kasutatud failitee või veebiaadressi. Kui jätkad tööd pärast kerneli taaskäivitamist, käivita esmalt see lahter uuesti.

## P2. Tabel, veerg ja andmete kontroll

**DataFrame** on ridadest ja veergudest koosnev andmetabel. Kui valid tabelist ühe veeru kujul `inimesed["vanus"]`, saad **Series**-objekti: väärtuste jada koos reaindeksiga.

**Indeks** annab igale reale tähise, mille järgi saab seda valida. Järgmises näites loob Pandas indeksiks arvud 0, 1, 2 ja 3. Indeks ei pruugi aga olla rea järjekorranumber: selleks võib kasutada ka teksti või kuupäevi. Filtreerimisel jäävad alles valitud ridade senised indeksiväärtused.

Hoia reaindeksi väärtused ja veerunimed unikaalsed ehk kordumatud. Tehniliselt lubab Pandas mõlemas kordusi, kuid neid tasub vältida: need võivad muuta valiku tulemuse ootamatuks või põhjustada mõnes toimingus vea.

Alusta väikese tabeliga. Sõnastiku võtmetest saavad veerunimed ja listidest veergude väärtused. Kõigis listides peab olema sama palju elemente.

Dokumentatsioon: [DataFrame](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html), [Series](https://pandas.pydata.org/docs/reference/api/pandas.Series.html).

Andmetabeli uurimisel küsi kõigepealt: mida üks rida tähendab, millised veerud on olemas ning kas väärtused on sobivat tüüpi?

- `.head(n)` tagastab esimesed `n` rida. Nende kuvamiseks kasuta `display()` või jäta meetodikutse lahtri viimaseks avaldiseks.
- `.shape` annab enniku `(ridade arv, veergude arv)`. See on atribuut, mistõttu selle järele väljakutse sulge ei lisata.
- `.info()` näitab veergude tüüpe ja olemasolevate väärtuste arvu.
- `.describe()` annab siin arvuliste veergude statistilise kokkuvõtte.

Täisarve sisaldava veeru tüüp võib olla `int64`, ujukomaarve sisaldaval veerul `float64`. Tekstiveeru tüübinimi võib Pandase versioonist sõltuvalt olla `str`, `string` või `object`. Andmetüüp ei määra üksi tunnuse tähendust: arvuline annusetase võib analüüsis olla kategooria.

Dokumentatsioon: [head](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.head.html), [shape](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.shape.html), [info](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.info.html), [describe](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.describe.html).

## P3. Valimine, filtreerimine ja sortimine

`tabel["veerg"]` valib ühe veeru ja tagastab Series-objekti. `tabel[["a", "b"]]` valib veerud `"a"` ja `"b"` ning tagastab neist koosneva tabeli. Sisemised nurksulud moodustavad veerunimede listi.

`.loc[read, veerud]` võimaldab valida korraga nii ridu kui ka veerge. Enne koma määra valitavad read indeksi tähiste või tingimusega; pärast koma anna veerunimi või veerunimede list.

Tingimus `inimesed["vanus"] >= 30` annab iga rea kohta väärtuse `True` või `False`. Kui kasutad seda tingimust ridade valimiseks, jäävad alles ainult `True`-ga märgitud read. Näites valime vähemalt 30-aastaste inimeste nimed ja linnad.

Dokumentatsioon: [loc](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.loc.html).

````python
# Veergude valimine veerunimede listiga.
display(inimesed[["nimi", "vanus"]])

# Võrdlus: iga rea kohta saadakse True või False.
display(inimesed["vanus"] >= 30)

# loc: tingimusele vastavate ridade ja soovitud veergude valimine.
vahemalt_30 = inimesed.loc[inimesed["vanus"] >= 30, ["nimi", "linn"]]
display(vahemalt_30)
````

Veergude võrdlemisel saadud tingimusi ühenda märkidega `&` (ja) ja `|` (või). Märk `~` pöörab tingimuse vastupidiseks. Pane iga võrdlus sulgudesse. Pythoni `and` ja `or` ei sobi tervete tõeväärtusveergude ühendamiseks.

`.sort_values()` järjestab read valitud veeru järgi. `ascending=False` annab kahaneva järjekorra. Järgmises näites tagastab meetod uue tabeli; algtabel jääb samaks.

Dokumentatsioon: [sort_values](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.sort_values.html).

````python
# Filtreerimine: mõlemad tingimused peavad kehtima.
tartu_valik = inimesed.loc[
    (inimesed["linn"] == "Tartu") & (inimesed["vanus"] > 25),
    ["nimi", "vanus"],
]
display(tartu_valik)

# Sortimine: suurima muutusega read esimesena.
display(algandmed.sort_values("Diff", ascending=False).head(5))
````

## P4. Uus veerg ja puuduvad väärtused

`.copy()` loob tabelist koopia, mida saad muuta algtabelit muutmata. Uue veeru lisamiseks kirjuta omistamise vasakule poolele tabeli nimi ja nurksulgudesse uus veerunimi, näiteks `andmed["arvutatud_muutus"]`. Paremale poole kirjuta arvutus.

Kahe veeru lahutamisel leitakse siin iga rea jaoks pärast- ja enneskoori vahe. Eraldi Pythoni tsüklit pole vaja.

Dokumentatsioon: [copy](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.copy.html).

````python
# Töökoopia loomine algandmeid muutmata.
andmed = algandmed.copy()

# Arvuline veerg: pärast- ja enneskoori vahe.
andmed["arvutatud_muutus"] = andmed["Mem_Score_After"] - andmed["Mem_Score_Before"]

# Tõeväärtusveerg: kas skoor suurenes?
andmed["skoor_suurenes"] = andmed["Diff"] > 0
display(andmed[["Mem_Score_Before", "Mem_Score_After", "Diff", "arvutatud_muutus", "skoor_suurenes"]].head())
````

Puuduvate väärtuste loendamine koosneb kahest sammust:

1. `.isna()` annab iga tabeliväärtuse kohta tõeväärtuse: `True`, kui väärtus puudub, ja `False`, kui see on olemas.
2. Sellele järgnev `.sum()` liidab tõeväärtused igas veerus. `True` läheb arvesse ühena ja `False` nullina, seega saad iga veeru puuduvate väärtuste arvu.

Puuduv väärtus tähendab, et andmed on teadmata või sisestamata. See ei ole sama mis arv 0. Enne eemaldamist või täitmist selgita välja, millised read muutuksid ja kuidas see analüüsi mõjutab.

Allpool on eraldi näidistabel, kus üks vanus puudub. `.dropna(subset=["vanus"])` jätab kõrvale ainult puuduva vanusega read. Seda võib kasutada analüüsis, mis vajab teadaolevat vanust; arvesta, et osa kirjeid jääb siis analüüsist välja.

Dokumentatsioon: [isna](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.isna.html), [sum](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.sum.html), [dropna](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.dropna.html).

````python
# isna() märgib puuduvad väärtused; sum() loendab need veergude kaupa.
display(algandmed.isna().sum())

# Eraldi näidistabel, milles üks vanus puudub.
naide = pd.DataFrame({"nimi": ["Mari", "Jaan"], "vanus": [24, None]})
display(naide)

# dropna(): puuduva vanusega ridade väljajätmine.
display(naide.dropna(subset=["vanus"]))
````

## P5. Rühmade kokkuvõte

`groupby("Drug")` koondab sama ravimikoodiga read ühte rühma. Sellele järgnev `.agg()` arvutab iga rühma kohta soovitud kokkuvõtted.

Kirjes `keskmine_muutus=("Diff", "mean")` on `keskmine_muutus` loodava veeru nimi. Sulgudes olev ennik määrab, millisest veerust väärtused võtta (`"Diff"`) ja milline arvutus teha (`"mean"` ehk keskmine).

- `mean` arvutab keskmise ja `median` mediaani. Puuduvad väärtused jäetakse neist arvutustest välja.
- `size` loendab kõik rühma read.
- `count` loendab valitud veeru olemasolevad väärtused. Puuduva väärtusega rida selle arvu sisse ei lähe.

Lisa keskmise kõrvale nii ridade arv kui ka kasutatud väärtuste arv. Nii näed rühma suurust ja seda, kui paljude väärtuste põhjal keskmine arvutati.

Dokumentatsioon: [groupby](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.groupby.html), [agg](https://pandas.pydata.org/docs/reference/api/pandas.api.typing.DataFrameGroupBy.agg.html), [size](https://pandas.pydata.org/docs/reference/api/pandas.api.typing.DataFrameGroupBy.size.html), [count](https://pandas.pydata.org/docs/reference/api/pandas.api.typing.DataFrameGroupBy.count.html).

````python
# groupby() ja agg(): rühmade kokkuvõtted.
kokkuvote = algandmed.groupby("Drug").agg(
    keskmine_muutus=("Diff", "mean"),
    mediaan=("Diff", "median"),
    ridu=("Diff", "size"),
    skooriga_ridu=("Diff", "count"),
).reset_index()  # Rühmitamistunnus tagasi tavaliseks veeruks.

# round(): ümardatud tulemuse kuvamine algset kokkuvõtet muutmata.
display(kokkuvote.round(2))
````

Pärast rühmitamist on ravimikood kokkuvõttetabeli indeksis. `.reset_index()` teeb sellest tavalise veeru `Drug` ning loob uue reaindeksi alates nullist.

`.round(2)` tagastab tabeli, mille arvväärtused on ümardatud kahe komakohani. Näites kuvame selle tulemuse; muutujas `kokkuvote` säilivad ümardamata väärtused.

Siin annavad `size` ja `count` sama tulemuse, sest veerus `Diff` pole puuduvaid väärtusi. Allolevas väikeses näites erinevad need arvud.

Dokumentatsioon: [reset_index](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.reset_index.html), [round](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.round.html).

````python
# Näidistabel: rühmas A on üks skoor puudu.
puuduvusega = pd.DataFrame({"ruhm": ["A", "A", "B"], "skoor": [5, None, 8]})

# size loendab kõik read; count ainult olemasolevad skoorid.
display(puuduvusega.groupby("ruhm").agg(ridu=("skoor", "size"), skooriga_ridu=("skoor", "count")))
````

### Rühmade läbimine for-tsükliga

`groupby()` tagastab `GroupBy`-objekti, mida saad läbida `for`-tsükliga. Igal sammul saad kaks väärtust:

- `ravimikood` on rühma võti, siin veeru `Drug` väärtus;
- `ruhma_tabel` on selle rühma kõiki ridu ja algseid veerge sisaldav `DataFrame`.

Näites kuvame iga ravimirühma suuruse ja kaks suurima skoorimuutusega kirjet. Sortimine ja ridade valimine toimuvad iga rühma sees eraldi.

`sort_values()` argument `key` määrab funktsiooni, mida rakendatakse veerule enne järjestamist. Siin kasutab `key=abs` absoluutväärtusi; tabelis säilivad algsed märgiga väärtused.

Tsükkel sobib siis, kui tahad iga rühmaga teha mitu toimingut, näiteks uurida selle ridu, koostada graafiku või salvestada eraldi faili. Tavalise kokkuvõttetabeli, näiteks rühmade keskmiste ja suuruste leidmiseks kasuta `.agg()`-i.

Dokumentatsioon: [rühmade läbimine](https://pandas.pydata.org/docs/user_guide/groupby.html#iterating-through-groups), [sort_values ja key](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.sort_values.html).

````python
# Rühmade läbimine: igal sammul rühma võti ja sellele vastav alamtabel.
for ravimikood, ruhma_tabel in algandmed.groupby("Drug"):
    print(f"Ravimirühm {ravimikood}: {len(ruhma_tabel)} kirjet")

    # Selle rühma kahe absoluutväärtuselt suurima muutuse valimine.
    suurimad = ruhma_tabel.sort_values("Diff", key=abs, ascending=False).head(2)
    display(suurimad[["Mem_Score_Before", "Mem_Score_After", "Diff"]])
````

## P7. Lisamaterjal: tabelite liitmine

`merge()` ühendab tabeleid ühise veeru ehk võtme väärtuste järgi. Näites lisame igale ravimikoodile ravimi nimetuse. `on="Drug"` määrab, et vasted leitakse veeru `Drug` järgi.

Kutses `algandmed.merge(ravimid, ...)` on vasakpoolne tabel `algandmed` ja parempoolne `ravimid`. `how="left"` jätab alles kõik vasakpoolse tabeli read, ka need, millele parempoolsest tabelist vastet ei leita. Sel juhul jääb lisatud raviminimetus puudu.

`validate="many_to_one"` kontrollib, et parempoolses tabelis esineb iga võtme väärtus ainult üks kord. Vasakpoolses tabelis võib sama ravimikood korduda. Kui parempoolses tabelis oleks sama kood mitmel real, võiks ühendamine suurendada tulemuse ridade arvu.

Salvesta tulemus uude muutujasse, et `algandmed` säiliks. Pärast ühendamist kontrolli ridade arvu ja seda, kas kõik raviminimetused on olemas. `.notna()` annab olemasoleva väärtuse korral `True`; sellele järgnev `.all()` kontrollib, kas kõik saadud tõeväärtused on `True`.

Dokumentatsioon: [merge](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.merge.html).

````python
# Vastavustabel: ravimikood ja nimetus.
ravimid = pd.DataFrame({
    "Drug": ["A", "T", "S"],
    "ravim": ["Alprasolaam", "Triasolaam", "Platseebo"],
})

# merge(): nimetuste ühendamine andmetega ravimikoodi järgi.
nimedega = algandmed.merge(ravimid, on="Drug", how="left", validate="many_to_one")

# Ühendamise kontroll: ridade arv säilis ja kõik nimetused leiti.
assert len(nimedega) == len(algandmed)
assert nimedega["ravim"].notna().all()
display(nimedega.head())
````

## P8. Lisamaterjal: lai ja pikk tabel

**Laias tabelis** on ühe katseisiku skoorid enne ja pärast katset sama rea eri veergudes: `Mem_Score_Before` ja `Mem_Score_After`.

**Pikas tabelis** on iga mõõtmine eraldi real. Veerg `skoor` sisaldab mõõdetud väärtust ja veerg `mootmine` näitab, kas see pärineb enne- või pärastmõõtmisest. Seega saab algtabeli ühest reast kaks rida. `.melt()` teeb selle teisenduse.

Enne teisendamist lisame töökoopiasse veeru `rea_id`, mis annab igale algtabeli reale kordumatu numbri. Selle järgi saad pärast teisendust kokku viia samalt algtabeli realt pärit mõõtmised. Katseisikute nimed korduvad, seega ei sobi nimi siin rea tuvastamiseks.

- `id_vars` loetleb veerud, mis jäävad mõlema mõõtmise juurde alles: siin `rea_id`, `Drug` ja `Dosage`.
- `value_vars` loetleb mõõtmisi sisaldavad veerud, mille väärtused lähevad ühte veergu.
- `var_name="mootmine"` annab nime veerule, kuhu lähevad algsed mõõtmisveergude nimed.
- `value_name="skoor"` annab nime veerule, kuhu lähevad skoorid.

Kontrollis rühmitame tulemuse `rea_id` järgi. `.size()` loendab iga rühma read, `.eq(2)` kontrollib, kas neid on kaks, ja `.all()` kontrollib, kas see kehtib kõigis rühmades.

Dokumentatsioon: [melt](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.melt.html).

````python
# Algsete ridade tähistamine enne tabeli teisendamist.
lai = algandmed.copy()
lai["rea_id"] = range(len(lai))

# melt(): kahe mõõtmisveeru teisendamine eraldi ridadeks.
pikk = lai.melt(
    id_vars=["rea_id", "Drug", "Dosage"],
    value_vars=["Mem_Score_Before", "Mem_Score_After"],
    var_name="mootmine",
    value_name="skoor",
)
display(pikk.head())
print("Ridu:", len(pikk))

# Kontroll: iga algse rea kohta on kaks mõõtmist.
assert pikk.groupby("rea_id").size().eq(2).all()
````

### Tabeli kuju muutmine pivot() abil

`pivot()` muudab ainult tabeli kuju. Iga rea- ja veerutunnuse kombinatsiooni kohta peab lähtetabelis olema kõige rohkem üks rida. Kui sama kombinatsioon kordub, kasuta kokkuvõtte arvutamiseks `pivot_table()`-it.

Järgmises näites muudame eelmise näite pika tabeli `pikk` tagasi laiaks. `index="rea_id"` määrab tulemuse read, `columns="mootmine"` veerud ja `values="skoor"` lahtritesse paigutatavad väärtused. Iga algse rea kohta on pikas tabelis üks enne- ja üks pärastmõõtmine. Seetõttu vastab igale `rea_id` ja `mootmine` kombinatsioonile üks skoor.

Tulemuses on enne- ja pärastmõõtmise skoorid jälle kõrvuti. Nende väärtused ei muutu ning keskmisi ei arvutata. Siin taastame ainult skooriveerud; `Drug` ja `Dosage` jäävad tulemusest välja.

Dokumentatsioon: [pivot](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.pivot.html).

````python
# pivot(): enne- ja pärastmõõtmise skoorid tagasi eri veergudesse.
skoorid_laias_tabelis = pikk.pivot(
    index="rea_id", columns="mootmine", values="skoor"
)
display(skoorid_laias_tabelis.head())
````

### Risttabeli koostamine pivot_table() abil

`pivot_table()` arvutab kokkuvõtte ja paigutab tulemused risttabelisse. Järgmises näites kasutame tabelit `algandmed`: ridadesse tulevad ravimirühmad (`index="Drug"`) ja veergudesse annusetasemed (`columns="Dosage"`). Igas lahtris on selle ravimirühma ja annusetaseme keskmine `Diff` (`values="Diff", aggfunc="mean"`).

Sama ravimirühma ja annusetaseme kohta on algandmetes mitu kirjet. `pivot()` annaks nende andmetega vea, sest ühte tulemuse lahtrisse peaks minema mitu väärtust. `pivot_table()` arvutab siin nendest väärtustest keskmise.

Dokumentatsioon: [pivot_table](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.pivot_table.html).

````python
# pivot_table(): keskmine muutus ravimirühma ja annusetaseme kaupa.
risttabel = algandmed.pivot_table(
    index="Drug", columns="Dosage", values="Diff", aggfunc="mean", observed=True
)

# Tulemuse ümardamine kuvamiseks.
display(risttabel.round(2))
````

`pd.cut()` määrab iga arvväärtuse etteantud piiride järgi rühma. `bins` määrab vahemike piirid ja `labels` rühmade nimetused.

Näites kuuluvad esimesse rühma vanused 0–35, teise üle 35 kuni 50, kolmandasse üle 50 kuni 65 ning viimasesse üle 65. `right=True` kaasab igasse vahemikku ülemise piiri; `include_lowest=True` kaasab esimesse vahemikku ka alumise piiri 0. `float("inf")` tähistab lõpmatust, nii et viimasel rühmal ei ole lõplikku ülempiiri.

Rühmade nimetused peavad vastama nende piiridele. Sildid `"36–50"`, `"51–65"` ja `"66+"` sobivad siin, sest vanused on täisaastates.

Dokumentatsioon: [cut](https://pandas.pydata.org/docs/reference/api/pandas.cut.html).

````python
# cut(): vanuste jagamine nimetatud rühmadesse.
vanustega = algandmed.copy()
vanustega["vanuseruhm"] = pd.cut(
    vanustega["age"],
    bins=[0, 35, 50, 65, float("inf")],
    labels=["kuni 35", "36–50", "51–65", "66+"],
    right=True,
    include_lowest=True,
)

# value_counts(): kirjete arvu leidmine igas vanuserühmas.
display(vanustega["vanuseruhm"].value_counts(sort=False))

# Kontroll: iga vanus sai rühma.
assert vanustega["vanuseruhm"].notna().all()
````

### Mitmest sammust koosnev töötlusahel

Töötlusahelas kasutab iga samm eelmise sammu tulemust. Tabeli meetodeid saad kutsuda järjest; `.pipe()` võimaldab lisada ahelasse ka eraldi funktsiooni.

Näites leiame **ravimirühmade keskmise skoori muutuse vähemalt 30-aastaste katseisikute seas** ja sordime tulemuse kahanevalt. Skoori muutus on algandmete veerus `Diff`.

Kirjutame funktsiooni `ruhmade_kokkuvote`, mis leiab iga rühma keskmise ja kirjete arvu. Rühmitamisveeru ja arvulise veeru nimed antakse argumentidena, seega saab sama funktsiooni kasutada ka teiste veergude või tabelitega. Mõlemad veerud peavad sisendtabelis olemas olema.

````python
# Kasutaja kirjutatud funktsioon: veerunimed antakse argumentidena.
def ruhmade_kokkuvote(tabel, ruhma_veerg, arvuline_veerg):
    return (
        tabel.groupby(ruhma_veerg)
        .agg(keskmine=(arvuline_veerg, "mean"), ridu=(arvuline_veerg, "size"))
        .reset_index()
    )
````

`.query()` valib tingimusele vastavad read; tingimus kirjutatakse siin tekstina. `.sort_values()` sordib tulemuse. Mõlemad on tabeli meetodid, mida kasutame ahelas otse.

`.pipe()` annab eelmise sammu tabeli funktsiooni `ruhmade_kokkuvote` esimeseks argumendiks ning edastab ka veerunimed. Funktsiooni nimi kirjutatakse ilma väljakutse sulgudeta: selle kutsub välja `.pipe()`. Iga sammu jaoks ei ole vaja oma funktsiooni kirjutada; `.pipe()`-ile sobib ka olemasolev teegifunktsioon.

Ahelat ümbritsevad sulud lubavad kirjutada iga sammu eraldi reale.

Dokumentatsioon: [query](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.query.html), [pipe](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.pipe.html).

````python
ahela_kokkuvote = (
    algandmed
    # 1. Tabeli meetod: vali vähemalt 30-aastased katseisikud.
    .query("age >= 30")
    # 2. pipe(): anna valitud read ja veerunimed meie kokkuvõttefunktsioonile.
    .pipe(ruhmade_kokkuvote, ruhma_veerg="Drug", arvuline_veerg="Diff")
    # 3. Tabeli meetod: sordi rühmad keskmise muutuse järgi kahanevalt.
    .sort_values("keskmine", ascending=False)
)
display(ahela_kokkuvote.round(2))
````
